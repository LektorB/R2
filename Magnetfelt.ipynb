```python
import pandas as pd
import numpy as np

# Les inn filen med pandas
# Vi bruker delimiter=';' for å skille kolonner og decimal=',' for norske tall
df = pd.read_csv('Spenning.csv', delimiter=';', decimal=',')

# Henter ut kolonnene basert på navnene i filen og lager numpy arrays
# Pandas håndterer automatisk headerne
tid = df['Latest: Time (s)'].to_numpy()
spenning = df['Latest: Potential (V)'].to_numpy()

# Skriver ut litt statistikk for å sjekke
print(f"Antall målepunkter: {len(tid)}")
print(f"Gjennomsnittlig spenning: {np.mean(spenning):.5f} V")
print(f"Maks spenning: {np.max(spenning):.5f} V")

# Skriver ut de første 5 verdiene
print("\nDe første 5 tidsverdiene:", tid[:5])
print("De første 5 spenningsverdiene:", spenning[:5])
```
