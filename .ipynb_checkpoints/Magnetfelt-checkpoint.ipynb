```python
import pandas as pd
import numpy as np
url = 'https://raw.githubusercontent.com/LektorB/R2/refs/heads/main/Spenning.csv'
# Les inn filen med pandas
# Vi bruker delimiter=';' for å skille kolonner og decimal=',' for norske tall
df = pd.read_csv(url, delimiter=';', decimal=',')

# Henter ut kolonnene basert på navnene i filen og lager numpy arrays
# Pandas håndterer automatisk headerne
tid = df['Latest: Time (s)'].to_numpy()
spenning = df['Latest: Potential (V)'].to_numpy()


print(f"Maks spenning: {np.max(spenning):.5f} V")
```

```python
plt.plot(tid,spenning)
```
